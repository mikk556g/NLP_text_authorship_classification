Load the contents of a URL or file, and strip out HTML markup:

```python
import nltk, re, pprint
from nltk import word_tokenize
from urllib import request
from bs4 import BeautifulSoup
```

```python
raw = open('text.txt').read()
type(raw)
```

```python
print(raw)
```

Tokenize a string which produces a list (of words):

```python
tokens = word_tokenize(raw)
type(tokens)
```

```python
print(tokens)
```

Normalizing and sorting the lists which produces another lists:

```python
words = [w.lower() for w in tokens]
type(words)
```

```python
vocab = sorted(set(words))
print(vocab)
```
