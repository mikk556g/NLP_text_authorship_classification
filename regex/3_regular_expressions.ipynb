Substitutions and Capture Groups

```python

# An important use of regular expressions is in substitutions, where we want to replace one string with another.
# In Python we use the function re.sub()


re.sub(pattern, repl, string) takes three arguments: 
# - a pattern to search for
# - a replacement to replace it with
# - and a string in which to do the search and replacing


# We could for example change every instance of "cherry" to "apricot" in string:
re.sub(r"cherry", r"apricot", string)

# Or we could convert to upper case all the instances of a particular name:
re.sub(r"janet", r"Janet", string)
```
