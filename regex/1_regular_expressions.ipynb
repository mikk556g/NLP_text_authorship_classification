```python
import regex as re
```

```python
# Searching for the uppercase letter "M" in the string "Mary had a little lam"
string = "Mary had a little lam"
re.search("M", string)
```

```python
# Using square brackets [] to search for either lowercase "m" or uppercase "M"
re.search("[mM]", string)
```

"""Specifying a range of something, with square brackets [] and dash -"""

```python
# Searching for any number between 3 and 6
numbers = "from 6 to 8"
re.search("[3-6]", numbers)
```

```python
# Searching for any uppercase letters
re.search("[A-Z]", string)
```

Using square brackets [] and the caret ^ to specify what a character CANNOT be: 
(^ has to be the first character in [], otherwise it's treated not as a function but the litteral character ^)

```python
# Using [] and ^ to search for anything but lowercase letters
re.search("[^a-z]", string)
```

Using the question mark ? to search with optional elements, e.g. "koala" and "koalas"
? means “the preceding character or nothing”

```python
string2 = "the koala is very cute"
re.search("koalas?", string2)
```

```python
string3 = "what color is that?"
re.search("colou?r", string3)
```

The Kleene star * means ZERO or more occurrences of the immediately previous character or regular expression

```python
sheep = "baa! baaa! baaaa!"
re.search("baaa*", sheep)
```

```python
# using [] with * here means, zero or more occurrences of a’s OR b’s
string4 = "abba is a music group"
re.search("[ab]*", string4)

# This will match strings like aaaa or ababab, abba or bbbb, as well as the empty string.

# As mentioned, * means zero or more occurrences of the immediately previous character OR regular expression -->
# "[ab]" is a regular expression, so "[ab]*" means zero or more of the regular expression "[ab]".
```

The Kleene + means ONE or more occurrences of the immediately preceding character or regular expression

```python
re.search("[0-9]+", numbers)

# So re.search("[0-9]+") is the normal way to specify a sequence of digits.
# We could also specify the sheep language as re.search("baa+!")
```

Using explicit numbers as counters, by enclosing them in curly brackets.

```python
# The operator re.search("{3}") means exactly 3 occurrences of the previous character or expression
# So re.search("ax{10}z") will match 'a' followed by exactly 10 'x' followed by 'z'.
```

The period re.search(".") is a wildcard expression that matches any single character (except a newline).

```python
# The wildcard is often used together with the Kleene star * to mean “any string of characters”.

# Example: suppose we want to find any line in which a particular word, for example, rose, appears twice. 
# We can specify this with the regular expression r"rose.*rose" -->
# meaning two roses, with a sequence of zero or more characters (of any kind) between them.

string5 = "a kiss from a rose, given by rose"
re.search("rose.*rose", string5)
```

```python
# Another example:
string6 = "a kiss from a rose-red rose"
re.search("rose.*rose", string6)
```

```python
# Using [] to allow the second 'rose' to be with either small 'r' or capital 'R':
re.search("rose.*[rR]ose", string5)
```
