Anchors and Boundaries: Anchors are special characters that anchor regular expressions to particular places in a string. 

The caret ^ and dollar sign $

```python
# The most common anchors are the caret ˆ, which matches the start of a line
# The pattern re.search("ˆThe") matches the word 'The' only at the start of a line. 

# Thus, the caret ˆ has three uses: 
# - To match the start of a line
# - To indicate a negation inside of square brackets
# - And just to mean a caret
```

```python
# The dollar sign $ matches the end of a line. So the pattern ␣$ is a useful pattern for matching a space at the end of a line.

# And re.search("ˆThe dog\.$") matches a line that contains only the phrase "The dog." with a final period.

# Note that we have to use the backslash (\) in the prior example since we want the . to mean “period” and not the wildcard. 

# By contrast, the regular expression re.search("ˆThe dog.$") would match e.g.:
# - "The dog."
# - "The dog!"
# - "The dogo"
```

Other word boundaries: \b and \B

```python
# A "word character" in regex is any digit, underscore, or letter ([a-zA-Z0-9_]). Everything else: spaces, punctuation, symbols like $, is a "non-word character."

# A word boundary (\b) is the invisible spot between a word character and a non-word character (or the start/end of the string). 
# A non-boundary (\B) is anywhere that transition doesn't happen — i.e., between two word characters, or between two non-word characters.
```

```python
# Common uses of \b
# 1. Whole-word matching, so you don't get substring hits.
#    - Search for r"\bcat\b" instead of r"cat" if you want to match "cat" but not "category," "bobcat," or "concatenate." 
#    - Without the boundaries \b around it, "cat" matches inside all of those.

# 2. Finding a specific number or code, not a fragment of a bigger one.
#    - r"\b99\b" finds the standalone number 99, not the "99" hiding inside 299 or 1999.
```

```python
# Common uses for \B
# \B is useful when you want the opposite — a match that's guaranteed to be in the interior of a word, explicitly not at an edge.

# 1. Matching inside a word only.
#    - r"\Bcat\B" would match "cat" in "concatenate" (surrounded by letters on both sides) but not in "cat" alone or "category" (where one side is a boundary).

# 2. Suffix-style matching without full-word matching.
#    - Say you want to find "ing" only when it's stuck onto the end of a longer word, not standing alone: r"\Bing\b" matches "running" but not the standalone word "ing"
```

```python
# Difference between r"cat" and r"\Bcat\B":

# r"cat" matches the literal characters "c", "a", "t" appearing anywhere, with no regard for what's next to them. 

# r"\Bcat\B" matches those same three characters, but only when there's no word boundary immediately before the "c" and no word boundary immediately after the "t".
# This means "cat" has to be sitting in the middle of a longer word-like sequence, with word characters directly touching it on both sides.


# Example:
# "cat" (the word by itself)

# r"cat" → matches. It doesn't care about context.

# r"\Bcat\B" → no match. Before the "c" is the start of the string, which counts as a boundary (edge → word character). \B requires not a boundary, so this fails immediately.
```

Disjunction, Grouping, and Precedence

```python
# The disjunction operator, also called the pipe symbol | is used to search for e.g. either "cat" or "dog"
# The pattern r"cat|dog" matches either the string cat or the string dog.


# Another example:
# How can you specify both "guppy" and "guppies"?

# We cannot simply say r"guppy|ies", because that would match only the strings "guppy" and "ies". 
# This is because sequences like "guppy" take precedence over the disjunction operator |.

# To make the disjunction operator apply only to a specific pattern, we need to use the parenthesis operators (). 
# Enclosing a pattern in parentheses makes it act like a single character for the purposes of neighboring operators like the pipe | and the *.

# So the pattern r"gupp(y|ies)" would specify that we meant the disjunction only to apply to the suffixes "y" and "ies".
```

```python
# The parenthesis operator () is also useful when we are using counters like the *. 
# Unlike the | operator, the * operator applies by default only to a single character, not to a whole sequence. 

# Suppose we want to match repeated instances of column labels of the form "Column 1", "Column 2", "Column 3". 
# The expression r"Column␣[0-9]+␣*" will not match our intended meaning of any number of Columns --> 
# instead, it will match a single instance of "Column" followed by any number of spaces.

# The * here applies only to the space ␣ that precedes it, not to the whole sequence. 
# With the parentheses (), we could write the expression r"(Column␣[0-9]+␣+)*" -->
# This matches the word "Column", followed by a number and spaces, and the whole pattern repeated zero or more times.
```

```python
"""
This idea that one operator may take precedence over another, requiring us to sometimes use parentheses to specify what we mean, 
is formalized by the operator precedence hierarchy for regular expressions. 

The following table gives the order of operator precedence, from highest precedence to lowest precedence:


1. Parenthesis                  ()
2. Counters                     * + ? {}
3. Sequences and anchors        the ^my end$
4. Disjunction                  |



"""
```

```python
# Patterns can be ambiguous in another way. Consider the expression r"[a-z]*" when matching against the text "once upon a time".
# Since r"[a-z]*" matches zero or more letters, this expression could match nothing, or just the first letter "o", "on", "onc", or "once".

# In these cases regular expressions always match the largest string they can; --> 
# we say that patterns are greedy, expanding to cover as much of a string as they can.


# There are, however, ways to enforce non-greedy matching, using another meaning of the ? qualifier.

# The operator *? is a Kleene * that matches as little text as possible.
# The operator +? is a Kleene + that matches as little text as possible.
```
